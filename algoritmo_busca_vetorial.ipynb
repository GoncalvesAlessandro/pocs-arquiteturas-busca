```python
!pip install sentence-transformers -q

import json
import time
import torch
from sentence_transformers import SentenceTransformer, util
from google.colab import drive

drive.mount('/content/drive')
caminho_ficheiro = '/content/drive/MyDrive/TCC_Base_Dados_POC/dataset_constituicao.json'

with open(caminho_ficheiro, 'r', encoding='utf-8') as f:
    artigos_cf = json.load(f)

textos_corpus = [doc['texto'] for doc in artigos_cf]

print("A descarregar e a carregar o modelo de Inteligência Artificial (Multilíngue)...")
modelo = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

print("\nA gerar os Embeddings (Vetores) para os 4.105 documentos...")

inicio_indexacao = time.perf_counter()

# gerando embedings
embeddings_corpus = modelo.encode(textos_corpus, convert_to_tensor=True)

fim_indexacao = time.perf_counter()
tempo_indexacao = fim_indexacao - inicio_indexacao

print(f"Base Vetorial pronta! \n⏱️ Tempo de indexação (Geração de Embeddings): {tempo_indexacao:.4f} segundos\n")
print("-----")

# Mesma consulta utilizada no BM25
consulta = "quais as regras para a eleição do presidente"

inicio_busca = time.perf_counter()

# Converte a pergunta num vetor e calcula a Similaridade de Cosseno com a base
embedding_consulta = modelo.encode(consulta, convert_to_tensor=True)
scores = util.cos_sim(embedding_consulta, embeddings_corpus)[0]

# Extrair o Top 3 com maior similaridade matemática
top_resultados = torch.topk(scores, k=3)

fim_busca = time.perf_counter()
tempo_busca = fim_busca - inicio_busca

print(f"Consulta: '{consulta}'")
print(f"Tempo de resposta da busca vetorial: {tempo_busca:.6f} segundos\n")

for i, (score, idx) in enumerate(zip(top_resultados.values, top_resultados.indices), 1):
    print(f"Top {i} (Similaridade: {score:.4f}):\n{textos_corpus[idx]}\n")
```
