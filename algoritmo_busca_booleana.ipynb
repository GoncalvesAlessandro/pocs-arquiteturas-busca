```python
import json
from google.colab import drive
import os

# load dataset
drive.mount('/content/drive')
caminho_ficheiro = '/content/drive/MyDrive/TCC_Base_Dados_POC/dataset_constituicao.json'

with open(caminho_ficheiro, 'r', encoding='utf-8') as f:
    artigos_cf = json.load(f)

print(f"Base carregada com {len(artigos_cf)} documentos.")

# definindo o motor de buusca
def busca_booleana(termos_and, termos_or, termos_not, documentos):
    resultados = []

    for doc in documentos:
        #normalizando textos para lowercase
        texto = doc['texto'].lower()

        # Lógica AND: O texto TEM de conter TODOS os termos desta lista
        match_and = all(termo.lower() in texto for termo in termos_and) if termos_and else True

        # Lógica OR: O texto TEM de conter PELO MENOS UM dos termos desta lista
        match_or = any(termo.lower() in texto for termo in termos_or) if termos_or else True

        # Lógica NOT: O texto NÃO PODE conter NENHUM dos termos desta lista
        match_not = not any(termo.lower() in texto for termo in termos_not) if termos_not else True

        # Para ser valido, precisa precisa passar em todos
        if match_and and match_or and match_not:
            resultados.append(doc['texto'])

    return resultados

# Run
print("----")
print("Executando Busca Booleana...\n")

# Cenário de Teste: Queremos saber sobre o "presidente" da "república",
# mas NÃO queremos resultados que falem sobre o "vice" ou "ministro".
termos_obrigatorios_and = ["presidente", "república"]
termos_opcionais_or = [] # Deixamos vazio neste teste
termos_proibidos_not = ["vice", "ministro"]

resultados = busca_booleana(termos_obrigatorios_and, termos_opcionais_or, termos_proibidos_not, artigos_cf)

print(f"Foram encontrados {len(resultados)} resultados que atendem perfeitamente à regra lógica.")
print("\nPrimeiros 3 resultados encontrados:\n")

for i, res in enumerate(resultados[:3], 1):
    print(f"Resultado {i}:\n{res}\n")
```
